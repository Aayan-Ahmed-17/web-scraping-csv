**Scrape data from website**

```python
import requests
from bs4 import BeautifulSoup
url = "https://freedium-mirror.cfd/"
blog_container_classes = "div.grid.w-full.grid-cols-1.gap-4.p-2"
item_card_classes = "div.p-6.bg-white.rounded-xl"

container = None

try:
  response = requests.get(url)
  soup = BeautifulSoup(response.content, "html.parser")
  container = soup.select_one(blog_container_classes)

except requests.exceptions.RequestException as e:
  print(f"Error In Request: {e}")

except Exception as e:
  print(f"Error: {e}")
```

**Print Data With Error Handling**

```python
if container is not None:
  try:
    item_cards = container.select(item_card_classes)
    print(f"Total cards: {len(item_cards)}")

    for i, item_card in enumerate(item_cards):
      img = item_card.select_one("img")
      heading = item_card.select_one("h3")
      description = item_card.select_one("p.mt-6.leading-normal.text-gray-600.line-clamp-3")
      if (img is not None) and (heading is not None) and (description is not None):
        print(i, img["src"])
        print(i, heading.text)
        print(i, description.text)
        print()
      else:
        print(i, "Content Not Found")

  except Exception as e:
    print(f"Error: {e}")
else:
  print("Container Data Not Found")
```

**Extract img_url, heading, desctiption & blog_url save them in variable**

```python
img = [item_card.select_one("img").get('src') for item_card in item_cards]
heading = [item_card.select_one("h3").text for item_card in item_cards]
description = [item_card.select_one("p.mt-6.leading-normal.text-gray-600.line-clamp-3").text for item_card in item_cards]
blog_url = ['https://freedium-mirror.cfd/'+ item_card.select_one("a.group").get('href') for item_card in item_cards]

print(img, heading, description, blog_url,sep='\n')
```

Save desired data in csv

```python
import csv
import os

if not os.path.exists('output'):
  os.makedirs('output')

with open('output/data.csv', mode='w', newline='') as fcsv:
  writer = csv.writer(fcsv)
  for data in zip(img, heading, description):
    writer.writerow(data)
```
